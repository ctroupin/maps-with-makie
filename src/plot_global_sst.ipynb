```julia
using NCDatasets
using Makie
using CairoMakie
using GeoMakie
```

```julia
sstfile = "/home/ctroupin/data/SST/SNPP_VIIRS.20260922_20260929.L3m.8D.SST3.sst_triple.4km.nc"
```

```julia
nc = NCDataset(sstfile, "r")
lonsst = coalesce(nc["lon"][:], NaN)
latsst = coalesce(nc["lat"][:], NaN)
sst = nc["sst_triple"][:,:]
close(nc)
```

```julia
fig = Figure(size = (1000, 1000))
ga1 = GeoAxis(fig[1, 1]; dest = "+proj=ortho +lon_0=-110 +lat_0=0")
surface!(ga1, lonsst, latsst, sst, shading = NoShading, interpolate = false, colormap=Reverse(:RdYlBu))
lines!(ga1, GeoMakie.coastlines(10), color=:black, linewidth=0.25)
hidedecorations!(ga1)
display(fig)
```
